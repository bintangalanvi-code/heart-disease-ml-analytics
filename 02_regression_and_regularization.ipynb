```python
# 1. Import Library yang Dibutuhkan (Sesuai Modul Coursera)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_predict, GridSearchCV
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score

import warnings
warnings.filterwarnings('ignore')

# 2. Membuat Dataset Simulasi Regresi Kesehatan (Biaya Perawatan Medis)
np.random.seed(72018)
n_samples = 500

age = np.random.randint(18, 75, size=n_samples)
bmi = np.random.uniform(18.5, 40.0, size=n_samples)
blood_pressure = np.random.randint(80, 180, size=n_samples)
lifestyle_score = np.random.uniform(1, 10, size=n_samples)

# Variabel Target: Biaya Perawatan Rumah Sakit (Medical Charges)
charges = 1000 + (age * 150) + (bmi * 200) + (blood_pressure * 50) - (lifestyle_score * 300) + np.random.normal(0, 500, size=n_samples)

df_medical = pd.DataFrame({
    'age': age,
    'bmi': bmi,
    'blood_pressure': blood_pressure,
    'lifestyle_score': lifestyle_score,
    'charges': charges
})

print("Dataset Medis Berhasil Dibuat!")
display(df_medical.head())

# 3. Memisahkan Fitur (X) dan Target (y)
X = df_medical.drop('charges', axis=1)
y = df_medical['charges']

# 4. Menerapkan Pipeline & Polynomial Features (Sesuai Materi Coursera)
# Menggabungkan Polynomial Features, StandardScaler, dan Ridge Regression dalam satu Pipeline
from sklearn.pipeline import Pipeline

estimator = Pipeline([
    ("polynomial_features", PolynomialFeatures(degree=2, include_bias=False)),
    ("scaler", StandardScaler()),
    ("ridge_regression", Ridge(alpha=1.0))
])

# 5. Train-Test Split (70% Train, 30% Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=72018)

# Fit model ke data training
estimator.fit(X_train, y_train)

# Prediksi dan Evaluasi Model
y_pred = estimator.predict(X_test)

print("\n=== EVALUASI MODEL REGRESI DENGAN PIPELINE & RIDGE ===")
print(f"R2 Score: {r2_score(y_test, y_pred):.4f}")
print(f"Root Mean Squared Error (RMSE): {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")

# 6. Hyperparameter Tuning menggunakan GridSearchCV (Materi Advanced Coursera)
params = {
    'polynomial_features__degree': [1, 2],
    'ridge_regression__alpha': np.geomspace(0.1, 10, 5)
}

grid = GridSearchCV(estimator, params, cv=3)
grid.fit(X, y)

print("\n=== HASIL GRID SEARCH HYPERPARAMETER TUNING ===")
print("Parameter Terbaik:", grid.best_params_)
print("Skor Terbaik (R2):", grid.best_score_)
```
